# Builds a clean cross-lingual dataset for ASKQE experiments.

Starting from English source sentences and automatically generated questions,
this script translates each context into Spanish (EN → ES) without introducing
intentional perturbations, producing a clean bilingual dataset used as a
reference condition in subsequent experiments.

```python
# IMPORT HERE
import json
import os
from tqdm import tqdm
from deep_translator import GoogleTranslator
from google.colab import drive

drive.mount('/content/drive')
```

```python
# FILE PATH
INPUT_QG_FILE = "QG/llama-8b/vanilla_llama-8b.jsonl"
OUTPUT_CLEAN_FILE = "QG/llama-8b/multilingual_clean_llama-8b.jsonl"

# parameter
MAX_EXAMPLES = None

print("⏳ Creazione dataset cross-lingual CLEAN (EN → ES)...")

# initialization of translators
translator_es = GoogleTranslator(source="en", target="es")
with open(INPUT_QG_FILE, "r", encoding="utf-8") as fin:
    lines = fin.readlines()
    if MAX_EXAMPLES is not None:
        lines = lines[:MAX_EXAMPLES]

with open(OUTPUT_CLEAN_FILE, "w", encoding="utf-8") as fout:
    for line in tqdm(lines):
        data = json.loads(line)

        en_text = data.get("en", "")

        if en_text:
            try:
                data["es"] = translator_es.translate(en_text)
            except Exception as e:
                # fallback in case of network error
                data["es"] = en_text
        else:
            data["es"] = ""

        fout.write(json.dumps(data, ensure_ascii=False) + "\n")

print(f"✅ Dataset cross-lingual CLEAN salvato in:\n   {OUTPUT_CLEAN_FILE}")
```
